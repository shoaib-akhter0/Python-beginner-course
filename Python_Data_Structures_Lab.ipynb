## 1. Python Lists

**What is a List?**
A list is an ordered, mutable collection that allows duplicate elements. Lists are created using square brackets `[]` and can contain elements of different data types.

```python
# Creating Lists
fruits = ["apple", "banana", "cherry", "date"]
print("Original list:", fruits)

# Accessing elements using index
print("First element:", fruits[0])
print("Last element:", fruits[-1])

# List operations
fruits.append("mango")
print("After append:", fruits)

fruits.remove("banana")
print("After remove:", fruits)

fruits.sort()
print("After sort:", fruits)

# List length
print("List length:", len(fruits))
```

## 2. Python Tuples

**What is a Tuple?**
A tuple is an ordered, immutable collection that allows duplicate elements. Tuples are created using parentheses `()` and cannot be modified after creation. This makes them useful for protecting data from accidental changes.

```python
# Creating Tuples
colors = ("red", "green", "blue", "yellow")
print("Original tuple:", colors)

# Accessing elements using index
print("First color:", colors[0])
print("Last color:", colors[-1])

# Tuple operations - concatenation
colors_extended = colors + ("orange", "purple")
print("Extended tuple:", colors_extended)

# Tuple operations - repetition
repeated_colors = ("white",) * 3
print("Repeated tuple:", repeated_colors)

# Tuple length
print("Tuple length:", len(colors))

# Why tuples are immutable
print("\nTrying to modify a tuple:")
try:
    colors[0] = "pink"
except TypeError as e:
    print("Error:", e)
```

## 3. Python Sets

**What is a Set?**
A set is an unordered, mutable collection with unique elements only. Sets are created using curly braces `{}` and automatically remove duplicates. Sets are useful for performing mathematical operations like union, intersection, and difference.

```python
# Creating Sets
numbers = {1, 2, 3, 4, 5}
print("Original set:", numbers)

# Duplicate handling
numbers_with_duplicates = {1, 2, 2, 3, 3, 3}
print("Set with duplicates removed:", numbers_with_duplicates)

# Adding elements
numbers.add(6)
print("After add:", numbers)

# Removing elements
numbers.remove(5)
print("After remove:", numbers)

# Set operations
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("\nSet Operations:")
print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference (set1 - set2):", set1.difference(set2))

# Set length
print("\nSet length:", len(set1))
```

## 4. Python Dictionaries

**What is a Dictionary?**
A dictionary is an unordered, mutable collection of key-value pairs. Dictionaries are created using curly braces `{}` with keys and values separated by colons. Dictionaries provide fast lookup and are useful for storing related data that needs to be accessed by a specific identifier.

```python
# Creating Dictionaries
student = {
    "name": "Alice",
    "age": 20,
    "major": "Computer Science",
    "gpa": 3.8
}
print("Original dictionary:", student)

# Accessing values using keys
print("Student name:", student["name"])
print("Student age:", student["age"])

# Adding new key-value pairs
student["email"] = "alice@example.com"
print("After adding email:", student)

# Updating values
student["age"] = 21
print("After updating age:", student)

# Removing key-value pairs
del student["major"]
print("After deleting major:", student)

# Iterating through dictionaries
print("\nIterating through dictionary:")
for key, value in student.items():
    print(f"  {key}: {value}")

# Dictionary methods
print("\nAll keys:", student.keys())
print("All values:", student.values())
print("Dictionary length:", len(student))
```
