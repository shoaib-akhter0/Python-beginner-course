# **Functions**

def function_name(parameters):

  code

  return value


def → used to create a function

function_name → name of the function

parameters → input values

return → sends a value back


```python
# Function definition
def greet():
    print("Hello, Shoaib!")
    print("Welcome to Python Functions.")

# Function call
greet()
```

# Function with a parameter

```python
def greet(name):
    print("Hello,", name)

greet("Shoaib")
greet("Ali")
```

# Function that returns a value

```python
def add_numbers(a, b):
    result = a + b
    return result

answer = add_numbers(5, 3)
print("Sum =", answer)
```

# Simple calculator function

```python
def multiply(num1, num2):
    return num1 * num2

result = multiply(4, 5)
print("Result =", result)
```

# Function to Print a Message

```python
def welcome():
    print("Welcome to Python!")

welcome()
```

# Function to Find Square of a Number

```python
def square(num):
    return num * num

result = square(5)
print("Square =", result)
```

# Function to Find the Larger Number

```python
def find_larger(a, b):
    if a > b:
        return a
    else:
        return b

print(find_larger(10, 20))
```

# Function to Check Even or Odd

```python
def check_even_odd(num):
    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")

check_even_odd(7)
```
