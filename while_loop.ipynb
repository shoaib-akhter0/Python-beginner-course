# **While loop**

A while loop is a control structure in Python that repeatedly executes a block of code as long as a specified condition remains true. It is useful when the number of iterations is not known in advance and the loop should continue until a certain condition is met.

Real-Life Examples

**ATM Machine** – The ATM keeps showing options while the user is logged in.

**Video Game** – The game runs while the player still has lives remaining.

**Mobile Phone Lock Screen** – The phone keeps asking for a password until the correct one is entered.

# 1. Print Numbers from 1 to 10

```python
count = 1

while count <= 10:
    print(count)
    count += 1
```

# 2. Print Even Numbers from 2 to 20

```python
num = 2

while num <= 20:
    print(num)
    num += 2
```

# 3. Print Multiplication Table of 5

```python
i = 1

while i <= 10:
    print(f"5 x {i} = {5 * i}")
    i += 1
```

# 4. Find the Sum of Numbers from 1 to 100

```python
i = 1
total = 0

while i <= 100:
    total += i
    i += 1

print("Sum =", total)
```

# 5. Countdown Timer

```python
count = 10

while count >= 1:
    print(count)
    count -= 1

print("Blast Off!")
```

# 6. Print Square of Numbers from 1 to 10

```python
i = 1

while i <= 10:
    print(i, "->", i ** 2)
    i += 1
```
