# NumPy Example Notebook
This notebook demonstrates NumPy usage with import, array creation, operations, saving/loading, and display of results.

## 1. Section: Import NumPy
Import the NumPy library and verify it is available in the notebook environment.

```python
import numpy as np

print('NumPy imported successfully. Version:', np.__version__)
```

## 2. Section: Create NumPy Arrays
Create NumPy arrays with different shapes and data types, including 1D and 2D arrays.

```python
# 1D array of integers
array_1d = np.array([1, 2, 3, 4, 5])

# 2D array of floats
array_2d = np.array([[1.5, 2.5, 3.5], [4.5, 5.5, 6.5]])

# Array with a specific dtype
array_int16 = np.array([10, 20, 30], dtype=np.int16)

array_1d, array_2d, array_int16
```

## 3. Section: Perform Array Operations
Use NumPy operations such as addition, multiplication, and statistical functions on arrays.

```python
sum_result = array_1d + np.array([5, 5, 5, 5, 5])
product_result = array_2d * 2
mean_2d = np.mean(array_2d)
std_2d = np.std(array_2d)

sum_result, product_result, mean_2d, std_2d
```

## 4. Section: Save and Load Data
Save NumPy arrays to disk and load them back into the notebook using NumPy IO functions.

```python
np.save('array_1d.npy', array_1d)
np.savez('arrays.npz', array_1d=array_1d, array_2d=array_2d)

loaded_array_1d = np.load('array_1d.npy')
loaded_data = np.load('arrays.npz')
loaded_array_2d = loaded_data['array_2d']

loaded_array_1d, loaded_array_2d
```

## 5. Section: Display Results in Notebook
Display array outputs and summaries directly in the notebook cells.

```python
print('Array 1D:', array_1d)
print('Array 2D:')
print(array_2d)
print('Loaded Array 1D:', loaded_array_1d)
print('Loaded Array 2D:')
print(loaded_array_2d)
print('Array 2D shape:', array_2d.shape)
print('Array 2D dtype:', array_2d.dtype)
```
