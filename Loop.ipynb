```python
print("\nExample 1: Numbers from 1 to 5")

for i in range(1, 6):
    print(i)
```

```python
print("\nExample 2: Student Names")

students = ["Ali", "Ahmed", "Sara", "John"]

for student in students:
    print("Welcome", student)
```

```python
print("\nExample 3: Table of 2")

for i in range(1, 11):
    print("2 x", i, "=", 2 * i)
```

```python
print("\nExample 4: Sum of Numbers")

total = 0

for i in range(1, 6):
    total = total + i

print("Total Sum =", total)
```

```python
# Example 5: Even numbers

print("\nExample 5: Even Numbers")

for i in range(2, 11, 2):
    print(i)
```

```python
print("\nProgram 3: Count Vowels")

sentence = "Welcome to Python Programming"
vowels = "aeiouAEIOU"
count = 0

for char in sentence:
    if char in vowels:
        count += 1

print("Sentence:", sentence)
print("Total Vowels:", count)
```

```python
print("\nProgram 6: Star Pattern")

rows = 5

for i in range(1, rows + 1):
    print("*" * i)
```

```python
print("\nProgram 10: Student Marks System")

students = ["Ali", "Ahmed", "Sara"]
marks = [85, 92, 78]

for i in range(len(students)):
    print(students[i], "got", marks[i], "marks")
```
